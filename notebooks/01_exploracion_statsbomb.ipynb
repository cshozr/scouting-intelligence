```python
import pandas as pd
import requests

# 1. Conectamos directamente al GitHub de StatsBomb para leer un partido de muestra (Euro 2016)
# Usamos la URL completa y correcta del archivo JSON correspondiente a un partido real
url_partido = "https://raw.githubusercontent.com/statsbomb/open-data/master/data/events/15946.json"
response = requests.get(url_partido).json()

# 2. Convertimos los datos crudos en una tabla limpia de Pandas
df_eventos = pd.json_normalize(response)

# 3. Guardamos los datos en tu carpeta web 'data/raw/' de forma virtual para ver qué hay
print(f"¡Éxito! Hemos cargado un partido con {len(df_eventos)} eventos registrados.")
print("\n--- PRIMERAS 5 FILAS DE LA TABLA ---")
df_eventos.head()
```

```python
# Cuenta cuántos eventos hay de cada tipo (Pases, Tiros, Faltas, etc.)
df_eventos['type.name'].value_counts()
```

```python
# Ver las columnas relacionadas con los nombres de los jugadores
df_eventos[['player.id', 'player.name', 'team.name']].dropna().drop_duplicates().head(10)
```

```python
# Filtramos solo las filas que sean tiros al arco
tiros = df_eventos[df_eventos['type.name'] == 'Shot']

# Seleccionamos columnas clave del tiro (jugador, minuto, resultado y si usó el pie/cabeza)
columnas_interes = ['player.name', 'minute', 'second', 'shot.outcome.name', 'shot.body_part.name']
tiros[columnas_interes].head()
```
